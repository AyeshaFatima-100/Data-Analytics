### Top 5 Best Selling Products
a) Write Observation
b) (Bar Chart) Using Seaborn

```python
top_products = df.groupby("product")["quantity"].sum().sort_values(ascending=False).head(5)
top_products
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title("Top 5 Best Selling Products")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.show()
```

### Payment Method Distribution
a) Write Observation
b) (Pie Chart) Using Matplotlib

```python
payment_counts = df["payment_method"].value_counts()
payment_counts
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    payment_counts.values,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Method Distribution")
plt.show()
```

### Month-wise Sales
a) Write Observation
b) (Line Chart) Using Matplotlib

```python
monthly_sales = df.groupby("Month")["total"].sum()

monthly_sales
```

```python
monthly_sales = monthly_sales.sort_index()

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Month-wise Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(monthly_sales.index)
plt.grid(True)
plt.show()
```

### Show City wise sales
a) Replace City Names
b) Horizontal Bar Chart using Seaborn
c) Write Observation

```python
df.head(1)
```

```python
df["City"].unique()
```

```python
df[["City", "total"]].head(20)
```

```python
city_mapping = {
    1.0: "City 1",
    2.0: "City 2",
    3.0: "City 3",
    4.0: "City 4",
    5.0: "City 5",
    6.0: "City 6"
}

df["City"] = df["City"].replace(city_mapping)
```

```python
city_sales = df.groupby("City")["total"].sum().sort_values(ascending=True)

city_sales
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_sales.values,
    y=city_sales.index
)

plt.title("City-wise Sales")
plt.xlabel("Total Sales")
plt.ylabel("City")
plt.show()
```

### Show monthly Sales by Category
a) Side by side bar chart using matplotlib
b) Write Observation

```python
monthly_category_sales = df.groupby(
    ["Month", "category"]
)["total"].sum().unstack()

print(monthly_category_sales)
```

```python
monthly_category_sales.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Monthly Sales by Category")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.legend(title="Category")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
