```python
class Employee:
    def __init__(self,name,number,address,salary,post,joining_year,education,total_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.total_experience = total_experience
        
    def describe_employee(self):
        print(F"\n------ Employee Details ------")
        print(f"name: {self.name}")
        print(f"number: {self.number}")
        print(f"address: {self.address}")
        print(f"salary: {self.salary}")
        print(f"post: {self.post}")
        print(F"year of joining: {self.joining_year}")
        print(f"education: {self.education}")
        print(F"years of experience: {self.total_experience}")
        
    def update_year_of_experience(self, years):
        self.total_experience += years
        print(f"{self.name}'s experience updated to {self.total_experience} years.")

    def increase_salary(self, increment):
        self.salary += increment
        print(f"{self.name}'s new salary is {self.salary}.")
        
employee101 = Employee("ayesha", 31119, "karachi", 75000, "staff tech", 2022, "BS", 6)


```

#### Question 2

Complete the same task (Question 1) without using the Classes or OOP.

```python
def describe_employee(employee):
    print("\n------ Employee Details ------")
    print("Name:", employee["name"])
    print("Employee Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", employee["salary"])
    print("Post:", employee["post"])
    print("Year of Joining:", employee["joining_year"])
    print("Education:", employee["education"])
    print("Years of Experience:", employee["experience"])


def update_year_of_experience(employee, years):
    employee["experience"] += years
    print(employee["name"], "experience updated to",
          employee["experience"], "years.")
    
def increase_salary(employee, amount):
    employee["salary"] += amount
    print(employee["name"], "salary increased to",
          employee["salary"])


employee101 = {
    "name": "Ali",
    "number": 101,
    "address": "Karachi",
    "salary": 50000,
    "post": "Manager",
    "joining_year": 2020,
    "education": "MBA",
   "experience": 5}
```

```python
describe_employee(employee101)
```

```python
update_year_of_experience(employee101, 2)
```

```python
increase_salary(employee101, 10000)
```

```python
describe_employee(employee101)
```
